### Vergleiche und Logik

Ein Vergleich ergibt immer `True` oder `False`.

| Operator | Bedeutung |
|---|---|
| `==` | gleich |
| `!=` | ungleich |
| `<`, `>` | kleiner, größer |
| `<=`, `>=` | kleiner gleich, größer gleich |

**Achtung:** `=` weist einen Wert zu. `==` vergleicht zwei Werte.

````python
print(x > y)    # -> True
print(x == 592) # -> True
print(x != y)   # -> True
````

Mit `and`, `or` und `not` verknüpfst du Wahrheitswerte:

- `and` → `True`, wenn **beide** Bedingungen wahr sind
- `or` → `True`, wenn **mindestens eine** Bedingung wahr ist
- `not` → kehrt den Wert um

````python
print(True and False)       # -> False
print(True or False)        # -> True
print(not False)            # -> True
print(x > 0 and y > 100)    # -> False
````

### Bedingungen: `if`, `elif`, `else`

Mit `if` führst du Code nur aus, wenn eine Bedingung wahr ist.

- Nach der Bedingung steht ein **Doppelpunkt** `:`.
- Der Code darunter ist **eingerückt** (4 Leerzeichen oder `Tab`).
- Die Einrückung zeigt Python, was zur Bedingung gehört.

````python
temperatur = 18.5

if temperatur > 25:
    print('Es ist heiß.')
elif temperatur > 10:
    print('Es ist mild.')
else:
    print('Es ist kalt.')
````

### Dictionaries

Ein Dictionary speichert **Paare aus Schlüssel und Wert**.
Du schreibst es in geschweiften Klammern `{}`.
Über den Schlüssel greifst du auf den Wert zu.

Beispiel: einige Bänder des Satelliten Sentinel-2 und ihre Wellenlänge in nm.

````python
sentinel2_bands = {'B02': 490, 'B03': 560, 'B04': 665, 'B08': 842}

print(sentinel2_bands['B04'])      # Wert zum Schlüssel 'B04'

sentinel2_bands['B11'] = 1610      # neues Paar hinzufügen

print(sentinel2_bands.keys())      # alle Schlüssel
print(sentinel2_bands.values())    # alle Werte
````

Die Formel lautet:

$f = c \cdot 1.8 + 32$

*Hintergrund:* Fahrenheit wollte negative Werte vermeiden.
Darum legte er den Nullpunkt auf die kälteste Temperatur, die er herstellen konnte.
Dazu mischte er Eis, Wasser und Salmiak.
Dieser Punkt liegt bei –17,8 °C. Das entspricht 0 °F.

````python
def cel_to_f(cel):
    f = cel*1.8+32
    return f

celsius = [0, 25, 100]

for c in celsius:
    print(cel_to_f(c))
````

**Lösung:** Versuche es erst selbst! Klicke dann links auf die Zelle, um den Code anzuzeigen.

````python
def celsius_to_fahrenheit(celsius):
    fahrenheit = celsius * 1.8 + 32
    return fahrenheit

print(celsius_to_fahrenheit(0))    # -> 32.0  (Gefrierpunkt von Wasser)
print(celsius_to_fahrenheit(25))   # -> 77.0  (warmer Sommertag)
print(celsius_to_fahrenheit(100))  # -> 212.0 (Siedepunkt von Wasser)
````

**Rechnen mit Arrays**

Eine Rechnung gilt automatisch für **jedes Element**.
Du brauchst dafür keine Schleife.

````python
temperaturen_c = np.array([12.1, 15.4, 18.9, 21.3, 17.0])

temperaturen_k = temperaturen_c + 273.15  # jedes Element + 273.15
print(temperaturen_k)

# Statistik
print(temperaturen_c.mean())  # Mittelwert
print(temperaturen_c.min())   # Minimum
print(temperaturen_c.max())   # Maximum
print(temperaturen_c.std())   # Standardabweichung
````

````python
maske = temperaturen_c > 16
print(maske)

print(temperaturen_c[maske])  # nur Werte über 16 °C
````

**Fehlende Werte: `np.nan`**

In echten Daten fehlen oft Werte, z. B. wegen Wolken.
Sie werden als `np.nan` (*Not a Number*) gespeichert.
Normale Statistik-Funktionen ergeben dann `nan`.
Nutze stattdessen `np.nanmean()`, `np.nanmax()` usw.

````python
messung = np.array([0.4, 0.6, np.nan, 0.5])

print(np.mean(messung))     # -> nan
print(np.nanmean(messung))  # ignoriert nan -> 0.5
````

**2D-Arrays: Rasterdaten**

Ein Satellitenbild ist ein 2D-Array.
Jedes Element ist ein Pixel.
Die erste Dimension sind die Zeilen, die zweite die Spalten.

Beispiel: Wir berechnen den **NDVI** (*Normalized Difference Vegetation Index*).
Er zeigt, wie viel grüne Vegetation in einem Pixel ist.

$NDVI = \frac{NIR - Rot}{NIR + Rot}$

Die Werte liegen zwischen -1 und 1.
Hohe Werte bedeuten dichte, gesunde Vegetation.

````python
# Reflexion im roten und im nahen Infrarot-Kanal (kleines Beispielbild mit 3 x 4 Pixeln)
rot = np.array([[0.05, 0.06, 0.20, 0.25],
                [0.04, 0.05, 0.18, 0.22],
                [0.03, 0.08, 0.15, 0.30]])

nir = np.array([[0.45, 0.50, 0.25, 0.28],
                [0.48, 0.42, 0.22, 0.24],
                [0.50, 0.40, 0.20, 0.31]])

print(rot.shape)  # (Zeilen, Spalten)
print(rot[0, 2])  # Pixel in Zeile 0, Spalte 2

ndvi = (nir - rot) / (nir + rot)  # Rechnung für jedes Pixel
ndvi.round(2)
````

### Matplotlib

Mit Matplotlib stellst du Daten grafisch dar.
Viele Beispiele findest du in der Galerie: [https://matplotlib.org/stable/gallery/index.html](https://matplotlib.org/stable/gallery/index.html)

Wir beenden jede Grafik mit `plt.show()`.
Ohne `plt.show()` erscheint die Grafik in Jupyter meist trotzdem.
Dann steht aber oft noch eine Textzeile darüber, z. B. `[<matplotlib.lines.Line2D ...>]`.

````python
x = np.arange(0, 10, 0.2)  # Werte von 0 bis 10 mit Abstand 0.2
y = np.sin(x)              # Sinus für jedes Element von x

plt.plot(x, y)             # Linie: y gegen x
plt.show()
````

Eine gute Grafik hat **Achsenbeschriftungen** und einen **Titel**.

````python
plt.plot(x, y)
plt.xlabel('x')            # Beschriftung der x-Achse
plt.ylabel('sin(x)')       # Beschriftung der y-Achse
plt.title('Meine erste Grafik')
plt.show()
````

Mehrere Linien zeichnest du in dieselbe Grafik.
Mit `label` und `plt.legend()` bekommst du eine **Legende**.

````python
z = np.cos(x)

plt.plot(x, y, label='sin(x)')
plt.plot(x, z, label='cos(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Meine zweite Grafik')
plt.legend()               # Legende anzeigen
plt.show()
````

Farbe und Linienart wählst du mit einem kurzen Text, z. B. `'b--'`:
`b` = blau, `--` = gestrichelt. `'r-'` heißt rot und durchgezogen.

Alle Optionen: [https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
plt.plot(x, y, 'b--', label='sin(x)')  # blau, gestrichelt
plt.plot(x, z, 'r-', label='cos(x)')   # rot, durchgezogen
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Meine dritte Grafik')
plt.legend()
plt.show()
````

Ein **Streudiagramm** (*scatter plot*) zeigt einzelne Punkte statt einer Linie.

````python
x = np.arange(0, 10, 0.5)  # Werte von 0 bis 10 mit Abstand 0.5
y = np.sin(x)

plt.scatter(x, y)
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Streudiagramm von sin(x)')
plt.show()
````

## Conda

> **Im Kurs ist keine eigene Conda-Umgebung nötig.**
> Wir arbeiten im VRE von VirES for Swarm.
> Dort sind alle wichtigen Packages schon installiert, z. B. `numpy`, `matplotlib`, `xarray`, `pandas` und `cartopy`.
>
> Trotzdem lohnt es sich zu wissen, wie Conda funktioniert.
> Spätestens bei einer Abschlussarbeit oder auf dem eigenen Laptop braucht man eigene Umgebungen.
> Dieser Abschnitt ist zum Lesen und Nachschlagen gedacht.

Conda verwaltet Packages und **Umgebungen**.
Eine Umgebung ist eine eigene Python-Installation für ein Projekt.

Warum ist das nützlich?
Packages wie `xarray`, `pandas` oder `cartopy` ändern sich ständig.
Nicht jede Version passt zu jeder anderen.
In einer eigenen Umgebung bleibt alles stabil.
Andere können die Umgebung genau nachbauen.

Mehr Infos:
- allgemein: [https://carpentries-incubator.github.io/introduction-to-conda-for-data-scientists/02-working-with-environments/index.html](https://carpentries-incubator.github.io/introduction-to-conda-for-data-scientists/02-working-with-environments/index.html)
- speziell für das VRE: [https://notebooks.vires.services/docs/vre-details](https://notebooks.vires.services/docs/vre-details)

**Wichtig im VRE:**
Nach einiger Zeit ohne Aktivität fährt dein VRE herunter.
Beim nächsten Start wird es neu aufgebaut.
Was du mit `pip install` oder `conda install` installiert hast, ist dann **weg**.
Nur dein **Home-Ordner** (`~`) bleibt erhalten.
Lege eigene Umgebungen deshalb immer im Home-Ordner an.

**So legst du eine Umgebung an:**

Öffne ein Terminal in JupyterLab (*File → New → Terminal*).

1. Neue Umgebung im Home-Ordner anlegen.
   Nach `--prefix` folgt der Ort der Umgebung.
   Danach folgen die Packages, die du brauchst.
   `ipykernel` ist nötig, damit Jupyter die Umgebung nutzen kann.

```bash
conda create --prefix ~/envs/awe_fernerkundung python=3.11 ipykernel xarray matplotlib zarr cartopy
```

2. Umgebung als Kernel in Jupyter verfügbar machen.
   Nach `--name` folgt ein interner Name.
   Nach `--display-name` folgt der Name, den du im Launcher siehst.

```bash
~/envs/awe_fernerkundung/bin/python -m ipykernel install --user --name awe_fernerkundung --display-name "AWE Fernerkundung"
```

3. Kernel auswählen: oben rechts im Notebook auf den Kernel-Namen klicken.
   Oder über *Kernel → Change Kernel*.

Später weitere Packages installieren:

```bash
conda activate ~/envs/awe_fernerkundung
conda install requests
conda deactivate
```

*Tipp:* `mamba` funktioniert wie `conda`, ist aber schneller.
